```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv(r"B:\Datos\Basal\Dataset_Reposo.csv")

display(data)
display()
```

```python
canales = ["AF3", "F7", "F3", "FC5", "FC6", "F4", "F8", "AF4", "P7", "O1", "O2", "P8"]
bandas  = ["Theta", "Alpha", "BetaL", "BetaH"]

columnas_filtradas = [f"POW.{canal}.{banda}" for canal in canales for banda in bandas]

dataf = data[columnas_filtradas]

display(dataf)
```

```python
dataf.info()
```

```python
Q1 = dataf.quantile(0.25)
Q3 = dataf.quantile(0.75)
IQR = Q3 - Q1

limite_inf = Q1 - 1.5 * IQR
limite_sup = Q3 + 1.5 * IQR

mask_outliers = (dataf < limite_inf) | (dataf > limite_sup)

filas_con_outliers = mask_outliers.any(axis=1)

print(f"Filas intervenidas: {filas_con_outliers.sum()}")

print("\n ANTES:")
display(dataf[filas_con_outliers])

df_limpio = dataf.copy()

for col in df_limpio.columns:
    mediana = df_limpio[col].median()
    df_limpio[col] = df_limpio[col].where(
        (df_limpio[col] >= limite_inf[col]) & (df_limpio[col] <= limite_sup[col]),
        other=mediana
    )

print("\n DESPUÉS:")
display(df_limpio[filas_con_outliers])
```

```python
df_limpio.to_csv(r"B:\Datos\Basal\Dataset_Reposo_filtrado.csv", index=False)
print("Guardado correctamente")
```
